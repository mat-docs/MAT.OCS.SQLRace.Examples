# Getting Started with SQL Race

This notebook walks through the basics: initialising the SQL Race runtime, loading a session, and reading your first parameter samples.

**Prerequisites:** Python 3.8+, `pythonnet`, ATLAS 10 installed.

## 1. Import helpers

The `sqlrace_helpers` module handles .NET runtime initialisation and provides convenient wrappers.

```python
import sys
sys.path.insert(0, '.')
from sqlrace_helpers import init_sqlrace, load_session, session_summary
```

## 2. Initialise the SQL Race runtime

This loads the .NET Core runtime and initialises the SQL Race library. It's safe to call multiple times.

```python
session_manager = init_sqlrace()
print("SQL Race initialised.")
```

## 3. Load a session

Replace the GUID below with a real session GUID from your database.

```python
SESSION_GUID = "<REPLACE WITH YOUR SESSION GUID>"

client_session, session = load_session(session_manager, SESSION_GUID)
session_summary(session)
```

## 4. List available parameters

Every session has a set of parameters (channels). Let's see what's available.

```python
from sqlrace_helpers import list_parameters

params = list_parameters(session)
print(f"Found {len(params)} parameter(s):")
for p in params[:10]:
    print(f"  {p}")
if len(params) > 10:
    print(f"  ... and {len(params) - 10} more")
```

## 5. Read parameter samples

Extract the first parameter as a pandas Series.

```python
from sqlrace_helpers import extract_parameter

if params:
    data = extract_parameter(session, params[0])
    print(f"Parameter: {params[0]}")
    print(f"Samples:   {len(data)}")
    print(f"\nFirst 5 values:")
    print(data.head())
else:
    print("No parameters in session.")
```

## 6. Quick plot

Visualise the parameter trace.

```python
import matplotlib.pyplot as plt

if params:
    fig, ax = plt.subplots(figsize=(12, 4))
    time_s = (data.index - data.index[0]) / 1e9
    ax.plot(time_s, data.values, linewidth=0.8)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel(params[0])
    ax.set_title(f"Parameter trace: {params[0]}")
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## 7. Clean up

Always dispose the client session when done.

```python
client_session.Dispose()
print("Session closed.")
```
